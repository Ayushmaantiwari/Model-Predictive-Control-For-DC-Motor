```python

# ================================================================
# DC MOTOR SPEED CONTROL: P vs PD vs PI vs PID vs MPC
# Complete interactive simulation with constrained MPC
# ================================================================

import os
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display, clear_output, FileLink
from scipy.optimize import minimize
from scipy.linalg import expm

OUTPUT_DIR = os.getcwd()
CONTROLLERS = ["P", "PD", "PI", "PID", "MPC"]

# ================================================================
# 1. MOTOR PARAMETERS
# ================================================================

Ra_w = widgets.FloatSlider(
    value=1.0, min=0.1, max=10.0, step=0.1,
    description="Ra (ohm)", continuous_update=False
)
La_w = widgets.FloatSlider(
    value=0.5, min=0.01, max=2.0, step=0.01,
    description="La (H)", continuous_update=False
)
J_w = widgets.FloatSlider(
    value=0.01, min=0.001, max=0.2, step=0.001,
    description="Inertia J", continuous_update=False
)
b_w = widgets.FloatSlider(
    value=0.1, min=0.001, max=1.0, step=0.001,
    description="Friction b", continuous_update=False
)
Kt_w = widgets.FloatSlider(
    value=0.01, min=0.001, max=1.0, step=0.001,
    description="Torque Kt", continuous_update=False
)
Ke_w = widgets.FloatSlider(
    value=0.01, min=0.001, max=1.0, step=0.001,
    description="Back EMF Ke", continuous_update=False
)

# ================================================================
# 2. REFERENCE AND SIMULATION SETTINGS
# ================================================================

ref_w = widgets.FloatSlider(
    value=50.0, min=1.0, max=300.0, step=1.0,
    description="Initial speed", continuous_update=False
)
ref2_w = widgets.FloatSlider(
    value=80.0, min=1.0, max=300.0, step=1.0,
    description="Second speed", continuous_update=False
)
duration_w = widgets.FloatSlider(
    value=5.0, min=1.0, max=15.0, step=0.5,
    description="Duration (s)", continuous_update=False
)
points_w = widgets.IntSlider(
    value=300, min=100, max=800, step=100,
    description="Samples", continuous_update=False
)
ref_change_w = widgets.FloatSlider(
    value=2.5, min=0.5, max=14.0, step=0.5,
    description="Ref change (s)", continuous_update=False
)

# ================================================================
# 3. PHYSICAL CONSTRAINTS
# ================================================================

voltage_limit_w = widgets.FloatSlider(
    value=24.0, min=5.0, max=100.0, step=1.0,
    description="Max |V|", continuous_update=False
)
current_limit_w = widgets.FloatSlider(
    value=10.0, min=1.0, max=50.0, step=0.5,
    description="Max |I|", continuous_update=False
)
speed_limit_w = widgets.FloatSlider(
    value=120.0, min=10.0, max=400.0, step=5.0,
    description="Max speed", continuous_update=False
)

# ================================================================
# 4. MPC PARAMETERS
# ================================================================

Np_w = widgets.IntSlider(
    value=10, min=2, max=25, step=1,
    description="Prediction Np", continuous_update=False
)
Nc_w = widgets.IntSlider(
    value=3, min=1, max=10, step=1,
    description="Control Nc", continuous_update=False
)
q_w = widgets.FloatLogSlider(
    value=10.0, base=10, min=-1, max=3, step=0.1,
    description="Tracking Q", continuous_update=False
)
r_w = widgets.FloatLogSlider(
    value=0.05, base=10, min=-3, max=2, step=0.1,
    description="Move penalty R", continuous_update=False
)
u_w = widgets.FloatLogSlider(
    value=0.002, base=10, min=-5, max=1, step=0.1,
    description="Voltage penalty S", continuous_update=False
)

# ================================================================
# 5. LOAD DISTURBANCE
# ================================================================

disturbance_w = widgets.FloatSlider(
    value=0.02, min=0.0, max=1.0, step=0.01,
    description="Load torque (Nm)", continuous_update=False
)
disturbance_time_w = widgets.FloatSlider(
    value=3.0, min=0.5, max=14.0, step=0.5,
    description="Disturbance time", continuous_update=False
)
enable_disturbance_w = widgets.Checkbox(
    value=True, description="Enable load disturbance"
)

# ================================================================
# 6. INDIVIDUAL P, PD, PI, PID GAINS
# ================================================================

def gain_widget(value, description):
    return widgets.FloatText(
        value=value,
        description=description,
        layout=widgets.Layout(width="170px")
    )

p_kp_w = gain_widget(0.5, "P: Kp")
pd_kp_w = gain_widget(0.5, "PD: Kp")
pd_kd_w = gain_widget(0.01, "PD: Kd")

pi_kp_w = gain_widget(0.5, "PI: Kp")
pi_ki_w = gain_widget(2.0, "PI: Ki")

pid_kp_w = gain_widget(0.5, "PID: Kp")
pid_ki_w = gain_widget(2.0, "PID: Ki")
pid_kd_w = gain_widget(0.01, "PID: Kd")

# ================================================================
# 7. GUI BUTTONS AND OUTPUT AREAS
# ================================================================

run_button = widgets.Button(
    description="Run all simulations",
    button_style="success",
    icon="play",
    layout=widgets.Layout(width="220px", height="42px")
)
reset_button = widgets.Button(
    description="Reset defaults",
    icon="refresh",
    layout=widgets.Layout(width="180px", height="42px")
)

status_output = widgets.Output()
plot_output = widgets.Output()
results_output = widgets.Output()
download_output = widgets.Output()

# ================================================================
# 8. CONTINUOUS-TO-DISCRETE MOTOR MODEL
# ================================================================

# State x = [armature current, angular speed]
#
# di/dt = (-Ra*i - Ke*w + V)/La
# dw/dt = (Kt*i - b*w - load torque)/J

def make_discrete_model(Ra, La, J, b, Kt, Ke, dt):
    A = np.array([
        [-Ra / La, -Ke / La],
        [Kt / J, -b / J]
    ], dtype=float)

    B = np.array([
        [1.0 / La],
        [0.0]
    ])

    E = np.array([
        [0.0],
        [-1.0 / J]
    ])

    augmented = np.zeros((4, 4))
    augmented[:2, :2] = A
    augmented[:2, 2:3] = B
    augmented[:2, 3:4] = E

    discrete = expm(augmented * dt)

    Ad = discrete[:2, :2]
    Bd = discrete[:2, 2:3]
    Ed = discrete[:2, 3:4]

    return Ad, Bd, Ed, A, B, E


def reference_at(t):
    if t < ref_change_w.value:
        return float(ref_w.value)
    return float(ref2_w.value)


def load_at(t):
    if enable_disturbance_w.value and t >= disturbance_time_w.value:
        return float(disturbance_w.value)
    return 0.0


# ================================================================
# 9. CONSTRAINED MODEL PREDICTIVE CONTROL
# ================================================================

def mpc_control(x, previous_voltage, k, t, dt, Ad, Bd, Ed, times):
    Np = int(Np_w.value)
    Nc = min(int(Nc_w.value), Np)

    vmax = float(voltage_limit_w.value)
    imax = float(current_limit_w.value)
    wmax = float(speed_limit_w.value)

    Q = float(q_w.value)
    R = float(r_w.value)
    S = float(u_w.value)

    initial = np.full(
        Nc, np.clip(previous_voltage, -vmax, vmax)
    )

    future_refs = np.array([
        min(reference_at(times[min(k + j, len(times) - 1)]), wmax)
        for j in range(1, Np + 1)
    ])

    future_loads = np.array([
        load_at(times[min(k + j, len(times) - 1)])
        for j in range(1, Np + 1)
    ])

    def rollout(u_sequence):
        xp = np.asarray(x, dtype=float).copy()
        states = []
        inputs = []

        for j in range(Np):
            uj = u_sequence[min(j, Nc - 1)]

            xp = (
                Ad @ xp
                + Bd[:, 0] * uj
                + Ed[:, 0] * future_loads[j]
            )

            states.append(xp.copy())
            inputs.append(uj)

        return np.asarray(states), np.asarray(inputs)

    def objective(u_sequence):
        states, inputs = rollout(u_sequence)

        speed = states[:, 1]
        voltage_changes = np.diff(
            np.r_[previous_voltage, inputs]
        )

        tracking_cost = Q * np.sum((future_refs - speed) ** 2)
        move_cost = R * np.sum(voltage_changes ** 2)
        effort_cost = S * np.sum(inputs ** 2)

        return tracking_cost + move_cost + effort_cost

    def state_constraints(u_sequence):
        states, _ = rollout(u_sequence)

        currents = states[:, 0]
        speeds = states[:, 1]

        # SLSQP inequality constraints must be >= 0.
        return np.r_[
            imax - currents,
            imax + currents,
            speeds,
            wmax - speeds
        ]

    bounds = [(-vmax, vmax)] * Nc

    try:
        solution = minimize(
            objective,
            initial,
            method="SLSQP",
            bounds=bounds,
            constraints=[
                {"type": "ineq", "fun": state_constraints}
            ],
            options={
                "maxiter": 35,
                "ftol": 1e-4,
                "disp": False
            }
        )

        if solution.success and np.all(np.isfinite(solution.x)):
            return float(solution.x[0]), True

        # Bounded fallback if optimization fails.
        error = reference_at(t) - x[1]
        fallback = previous_voltage + 0.5 * error

        return float(np.clip(fallback, -vmax, vmax)), False

    except Exception:
        error = reference_at(t) - x[1]
        fallback = previous_voltage + 0.5 * error

        return float(np.clip(fallback, -vmax, vmax)), False


# ================================================================
# 10. CLASSICAL P, PD, PI, PID CONTROLLERS
# ================================================================

def classical_control(
    name, x, integral, previous_error,
    previous_voltage, dt, derivative_state
):
    error = reference_at(current_time) - x[1]

    # Derivative of measured speed reduces derivative kick when
    # the reference changes abruptly.
    measured_speed_derivative = (
        (x[1] - derivative_state) / dt if dt > 0 else 0.0
    )

    if name == "P":
        kp, ki, kd = float(p_kp_w.value), 0.0, 0.0

    elif name == "PD":
        kp, ki, kd = (
            float(pd_kp_w.value), 0.0, float(pd_kd_w.value)
        )

    elif name == "PI":
        kp, ki, kd = float(pi_kp_w.value), float(pi_ki_w.value), 0.0

    else:  # PID
        kp, ki, kd = (
            float(pid_kp_w.value),
            float(pid_ki_w.value),
            float(pid_kd_w.value)
        )

    vmax = float(voltage_limit_w.value)

    raw = (
        kp * error
        + ki * integral
        - kd * measured_speed_derivative
    )

    output = float(np.clip(raw, -vmax, vmax))

    # Conditional integration for basic anti-windup.
    if (
        abs(raw - output) < 1e-9
        or (raw > vmax and error < 0)
        or (raw < -vmax and error > 0)
    ):
        integral += error * dt

    raw = (
        kp * error
        + ki * integral
        - kd * measured_speed_derivative
    )

    output = float(np.clip(raw, -vmax, vmax))

    return output, integral, error


# ================================================================
# 11. SIMULATE ONE CONTROLLER
# ================================================================

def simulate_controller(name, times, dt, Ad, Bd, Ed):
    x = np.zeros(2, dtype=float)

    previous_voltage = 0.0
    integral = 0.0
    previous_error = 0.0
    derivative_state = 0.0

    records = []
    optimizer_success = []

    global current_time

    for k, t in enumerate(times):
        current_time = float(t)

        reference = min(
            reference_at(t), float(speed_limit_w.value)
        )
        load = load_at(t)

        if name == "MPC":
            voltage, success = mpc_control(
                x, previous_voltage, k, t, dt,
                Ad, Bd, Ed, times
            )
            optimizer_success.append(success)

        else:
            voltage, integral, error = classical_control(
                name,
                x,
                integral,
                previous_error,
                previous_voltage,
                dt,
                derivative_state
            )

            previous_error = error

        voltage = float(np.clip(
            voltage,
            -float(voltage_limit_w.value),
            float(voltage_limit_w.value)
        ))

        records.append({
            "time_s": t,
            "controller": name,
            "reference_rad_s": reference,
            "speed_rad_s": x[1],
            "current_A": x[0],
            "voltage_V": voltage,
            "load_torque_Nm": load,
            "optimizer_success": (
                optimizer_success[-1] if name == "MPC" else True
            )
        })

        derivative_state = x[1]

        # Simulate the motor without artificially clipping its state.
        x = (
            Ad @ x
            + Bd[:, 0] * voltage
            + Ed[:, 0] * load
        )

        previous_voltage = voltage

    return pd.DataFrame(records)


# ================================================================
# 12. PERFORMANCE METRICS
# ================================================================

def calculate_metrics(df):
    ref = df["reference_rad_s"].to_numpy()
    speed = df["speed_rad_s"].to_numpy()
    voltage = df["voltage_V"].to_numpy()
    current = df["current_A"].to_numpy()
    t = df["time_s"].to_numpy()

    error = ref - speed

    rmse = float(np.sqrt(np.mean(error ** 2)))
    mae = float(np.mean(np.abs(error)))
    max_error = float(np.max(np.abs(error)))

    # Overshoot relative to the largest requested positive speed.
    overshoot = max(
        0.0,
        (float(np.max(speed)) - float(np.max(ref)))
        / max(float(np.max(ref)), 1e-9) * 100.0
    )

    # Settling time after the final reference change.
    final_ref = float(ref[-1])
    final_change_idx = np.where(np.diff(ref) != 0)[0]

    if len(final_change_idx):
        start_idx = int(final_change_idx[-1] + 1)
    else:
        start_idx = 0

    band = max(0.02 * abs(final_ref), 0.1)
    inside = np.abs(speed - final_ref) <= band

    settling_time = np.nan

    for idx in range(start_idx, len(t)):
        if np.all(inside[idx:]):
            settling_time = float(t[idx] - t[start_idx])
            break

    # Rise time to 90% of the first positive reference.
    first_ref = float(ref[0])
    first_segment = np.where(ref != first_ref)[0]

    if len(first_segment):
        end_idx = int(first_segment[0])
    else:
        end_idx = len(t)

    rise_time = np.nan

    if first_ref > 0:
        hits = np.where(speed[:end_idx] >= 0.9 * first_ref)[0]

        if len(hits):
            rise_time = float(t[hits[0]] - t[0])

    vmax = float(voltage_limit_w.value)
    imax = float(current_limit_w.value)
    wmax = float(speed_limit_w.value)

    voltage_violations = int(
        np.sum(np.abs(voltage) > vmax + 1e-5)
    )
    current_violations = int(
        np.sum(np.abs(current) > imax + 1e-5)
    )
    speed_violations = int(
        np.sum((speed < -1e-5) | (speed > wmax + 1e-5))
    )

    return {
        "Controller": df["controller"].iloc[0],
        "RMSE (rad/s)": rmse,
        "MAE (rad/s)": mae,
        "Maximum error (rad/s)": max_error,
        "Overshoot (%)": overshoot,
        "Rise time (s)": rise_time,
        "Settling time (s)": settling_time,
        "Peak |current| (A)": float(np.max(np.abs(current))),
        "Peak |voltage| (V)": float(np.max(np.abs(voltage))),
        "Voltage violations": voltage_violations,
        "Current violations": current_violations,
        "Speed violations": speed_violations,
        "MPC optimizer success (%)": (
            float(df["optimizer_success"].mean() * 100)
            if df["controller"].iloc[0] == "MPC"
            else np.nan
        )
    }


# ================================================================
# 13. MAIN SIMULATION FUNCTION
# ================================================================

def run_simulation(_=None):
    for output in [
        status_output, plot_output,
        results_output, download_output
    ]:
        with output:
            clear_output(wait=True)

    with status_output:
        print("Starting simulation for P, PD, PI, PID and MPC...")

    # Input validation
    if Nc_w.value > Np_w.value:
        with status_output:
            clear_output(wait=True)
            print("Error: Nc must be less than or equal to Np.")
        return

    if La_w.value <= 0 or J_w.value <= 0:
        with status_output:
            clear_output(wait=True)
            print("Error: Inductance and inertia must be positive.")
        return

    gain_values = [
        p_kp_w.value,
        pd_kp_w.value, pd_kd_w.value,
        pi_kp_w.value, pi_ki_w.value,
        pid_kp_w.value, pid_ki_w.value, pid_kd_w.value
    ]

    if not np.all(np.isfinite(gain_values)):
        with status_output:
            clear_output(wait=True)
            print("Error: Controller gains must be finite.")
        return

    if any(value < 0 for value in gain_values):
        with status_output:
            clear_output(wait=True)
            print("Error: This GUI expects nonnegative controller gains.")
        return

    duration = float(duration_w.value)
    n = int(points_w.value)

    times = np.linspace(0, duration, n)
    dt = float(times[1] - times[0])

    Ra, La, J, b, Kt, Ke = [
        float(w.value)
        for w in [Ra_w, La_w, J_w, b_w, Kt_w, Ke_w]
    ]

    Ad, Bd, Ed, A, B, E = make_discrete_model(
        Ra, La, J, b, Kt, Ke, dt
    )

    all_results = []
    metric_rows = []

    try:
        for name in CONTROLLERS:
            with status_output:
                clear_output(wait=True)
                print(f"Simulating {name} controller...")

            result = simulate_controller(
                name, times, dt, Ad, Bd, Ed
            )

            all_results.append(result)
            metric_rows.append(calculate_metrics(result))

        full_df = pd.concat(all_results, ignore_index=True)
        metrics_df = pd.DataFrame(metric_rows)

        # --------------------------------------------------------
        # FIGURE 1: SPEED, VOLTAGE, CURRENT, TRACKING ERROR
        # --------------------------------------------------------

        fig, axes = plt.subplots(2, 2, figsize=(16, 10))

        for name in CONTROLLERS:
            d = full_df[full_df["controller"] == name]

            axes[0, 0].plot(
                d["time_s"], d["speed_rad_s"],
                label=name, linewidth=1.6
            )

            axes[0, 1].plot(
                d["time_s"], d["voltage_V"],
                label=name, linewidth=1.3
            )

            axes[1, 0].plot(
                d["time_s"], d["current_A"],
                label=name, linewidth=1.3
            )

            axes[1, 1].plot(
                d["time_s"],
                d["reference_rad_s"] - d["speed_rad_s"],
                label=name, linewidth=1.3
            )

        ref_df = full_df[full_df["controller"] == "MPC"]

        axes[0, 0].plot(
            ref_df["time_s"], ref_df["reference_rad_s"],
            "k--", label="Reference", linewidth=1.8
        )
        axes[0, 0].axhline(
            speed_limit_w.value, color="red",
            linestyle=":", label="Speed limit"
        )

        axes[0, 1].axhline(
            voltage_limit_w.value, color="black", linestyle="--"
        )
        axes[0, 1].axhline(
            -voltage_limit_w.value, color="black", linestyle="--"
        )

        axes[1, 0].axhline(
            current_limit_w.value, color="black", linestyle="--"
        )
        axes[1, 0].axhline(
            -current_limit_w.value, color="black", linestyle="--"
        )

        axes[1, 1].axhline(0, color="black", linestyle="--")

        axes[0, 0].set_title("Speed Tracking")
        axes[0, 0].set_ylabel("Speed (rad/s)")
        axes[0, 1].set_title("Armature Voltage")
        axes[0, 1].set_ylabel("Voltage (V)")
        axes[1, 0].set_title("Armature Current")
        axes[1, 0].set_ylabel("Current (A)")
        axes[1, 1].set_title("Tracking Error")
        axes[1, 1].set_ylabel("Error (rad/s)")

        for ax in axes.flat:
            ax.set_xlabel("Time (s)")
            ax.grid(True, alpha=0.3)
            ax.legend()

        fig.suptitle(
            "DC Motor Control: P vs PD vs PI vs PID vs MPC"
        )
        fig.tight_layout()

        dashboard_path = os.path.join(
            OUTPUT_DIR, "dc_motor_mpc_dashboard.png"
        )
        fig.savefig(dashboard_path, dpi=180, bbox_inches="tight")

        with plot_output:
            display(fig)

        plt.close(fig)

        # --------------------------------------------------------
        # FIGURE 2: PERFORMANCE COMPARISON
        # --------------------------------------------------------

        fig2, axes2 = plt.subplots(2, 2, figsize=(14, 9))

        metrics_to_plot = [
            "RMSE (rad/s)",
            "Overshoot (%)",
            "Settling time (s)",
            "Peak |current| (A)"
        ]

        for ax, metric in zip(axes2.flat, metrics_to_plot):
            vals = metrics_df[metric].fillna(0)

            ax.bar(metrics_df["Controller"], vals)
            ax.set_title(metric)
            ax.grid(axis="y", alpha=0.3)

        fig2.suptitle("Controller Performance Comparison")
        fig2.tight_layout()

        metrics_plot_path = os.path.join(
            OUTPUT_DIR, "dc_motor_mpc_metrics.png"
        )
        fig2.savefig(
            metrics_plot_path, dpi=180, bbox_inches="tight"
        )

        with plot_output:
            display(fig2)

        plt.close(fig2)

        # --------------------------------------------------------
        # EXPORT CSV FILES
        # --------------------------------------------------------

        timeseries_path = os.path.join(
            OUTPUT_DIR, "dc_motor_mpc_timeseries.csv"
        )
        metrics_path = os.path.join(
            OUTPUT_DIR, "dc_motor_mpc_metrics.csv"
        )

        full_df.to_csv(timeseries_path, index=False)
        metrics_df.to_csv(metrics_path, index=False)

        # --------------------------------------------------------
        # WRITE SIMULATION REPORT
        # --------------------------------------------------------

        report_path = os.path.join(
            OUTPUT_DIR, "dc_motor_mpc_report.txt"
        )

        with open(report_path, "w", encoding="utf-8") as f:
            f.write("DC MOTOR SPEED CONTROL SIMULATION REPORT\n")
            f.write("=" * 52 + "\n\n")

            f.write("Controllers: P, PD, PI, PID, MPC\n\n")

            f.write("Motor parameters:\n")
            f.write(f"Ra = {Ra} ohm\n")
            f.write(f"La = {La} H\n")
            f.write(f"J = {J} kg m^2\n")
            f.write(f"b = {b} N m s\n")
            f.write(f"Kt = {Kt}\nKe = {Ke}\n")
            f.write(f"Sampling time = {dt:.6g} s\n")
            f.write(f"Duration = {duration} s\n\n")

            f.write("Reference settings:\n")
            f.write(f"Initial reference = {ref_w.value} rad/s\n")
            f.write(f"Second reference = {ref2_w.value} rad/s\n")
            f.write(f"Reference change time = {ref_change_w.value} s\n\n")

            f.write("Constraints:\n")
            f.write(f"Voltage = +/-{voltage_limit_w.value} V\n")
            f.write(f"Current = +/-{current_limit_w.value} A\n")
            f.write(f"Speed = 0 to {speed_limit_w.value} rad/s\n\n")

            f.write("MPC settings:\n")
            f.write(f"Np = {Np_w.value}\n")
            f.write(f"Nc = {Nc_w.value}\n")
            f.write(f"Q = {q_w.value}\n")
            f.write(f"R = {r_w.value}\n")
            f.write(f"S = {u_w.value}\n\n")

            f.write("Controller gains:\n")
            f.write(f"P: Kp = {p_kp_w.value}\n")
            f.write(
                f"PD: Kp = {pd_kp_w.value}, Kd = {pd_kd_w.value}\n"
            )
            f.write(
                f"PI: Kp = {pi_kp_w.value}, Ki = {pi_ki_w.value}\n"
            )
            f.write(
                f"PID: Kp = {pid_kp_w.value}, "
                f"Ki = {pid_ki_w.value}, Kd = {pid_kd_w.value}\n\n"
            )

            f.write("Disturbance:\n")
            f.write(f"Enabled = {enable_disturbance_w.value}\n")
            f.write(f"Load torque = {disturbance_w.value} Nm\n")
            f.write(
                f"Disturbance time = {disturbance_time_w.value} s\n\n"
            )

            f.write("Performance metrics:\n")
            f.write(metrics_df.round(5).to_string(index=False))

            f.write("\n\nInterpretation:\n")
            f.write(
                "P uses proportional action. PD adds derivative action "
                "to improve damping. PI adds integral action to reduce "
                "steady-state error. PID combines all three actions. "
                "MPC optimizes predicted tracking error and control effort "
                "subject to predicted state constraints.\n\n"
            )

            f.write(
                "Constraint violations must be checked explicitly. "
                "The MPC optimizer enforces predicted current and speed "
                "limits only when a feasible optimization solution is "
                "found. Its fallback voltage does not guarantee those "
                "limits. P, PD, PI and PID enforce voltage saturation "
                "but do not enforce hard current or speed constraints. "
                "The results are from a simulation, not a hardware test.\n"
            )

        # --------------------------------------------------------
        # DISPLAY METRICS AND MODEL
        # --------------------------------------------------------

        with results_output:
            display(metrics_df.round(4))

            print("\nDiscrete-time motor model:")
            print("x[k+1] = Ad*x[k] + Bd*V[k] + Ed*load[k]")

            print("\nAd =\n", np.round(Ad, 5))
            print("\nBd =\n", np.round(Bd, 5))
            print("\nEd =\n", np.round(Ed, 5))

            print("\nState vector: [armature current, angular speed]")
            print("Reference: piecewise constant speed command")

        # --------------------------------------------------------
        # DOWNLOAD LINKS
        # --------------------------------------------------------

        with download_output:
            print("Download simulation results:")

            for path in [
                timeseries_path,
                metrics_path,
                dashboard_path,
                metrics_plot_path,
                report_path
            ]:
                display(FileLink(path))

        with status_output:
            clear_output(wait=True)
            print(
                "Simulation completed for all five controllers. "
                "Review the graphs, metrics and constraint violations."
            )

    except Exception as exc:
        with status_output:
            clear_output(wait=True)
            print("Simulation error:", repr(exc))
            print(
                "Try reducing the sample count or prediction horizon, "
                "and check the controller gain values."
            )


# ================================================================
# 14. RESET ALL DEFAULTS
# ================================================================

def reset_defaults(_=None):
    defaults = {
        Ra_w: 1.0,
        La_w: 0.5,
        J_w: 0.01,
        b_w: 0.1,
        Kt_w: 0.01,
        Ke_w: 0.01,

        ref_w: 50.0,
        ref2_w: 80.0,
        duration_w: 5.0,
        points_w: 300,
        ref_change_w: 2.5,

        voltage_limit_w: 24.0,
        current_limit_w: 10.0,
        speed_limit_w: 120.0,

        Np_w: 10,
        Nc_w: 3,
        q_w: 10.0,
        r_w: 0.05,
        u_w: 0.002,

        disturbance_w: 0.02,
        disturbance_time_w: 3.0,
        enable_disturbance_w: True,

        p_kp_w: 0.5,
        pd_kp_w: 0.5,
        pd_kd_w: 0.01,

        pi_kp_w: 0.5,
        pi_ki_w: 2.0,

        pid_kp_w: 0.5,
        pid_ki_w: 2.0,
        pid_kd_w: 0.01
    }

    for widget, value in defaults.items():
        widget.value = value

    with status_output:
        clear_output(wait=True)
        print("Defaults restored. Click Run all simulations.")


run_button.on_click(run_simulation)
reset_button.on_click(reset_defaults)

# ================================================================
# 15. INTERACTIVE GUI
# ================================================================

motor_box = widgets.VBox([
    widgets.HTML("<h3>1. DC Motor Parameters</h3>"),
    widgets.HBox([
        widgets.VBox([Ra_w, La_w, J_w]),
        widgets.VBox([b_w, Kt_w, Ke_w])
    ])
])

simulation_box = widgets.VBox([
    widgets.HTML("<h3>2. Reference and Simulation Settings</h3>"),
    widgets.HBox([ref_w, ref2_w]),
    widgets.HBox([duration_w, points_w, ref_change_w])
])

constraints_box = widgets.VBox([
    widgets.HTML("<h3>3. Physical Constraints</h3>"),
    widgets.HBox([
        voltage_limit_w, current_limit_w, speed_limit_w
    ])
])

mpc_box = widgets.VBox([
    widgets.HTML("<h3>4. MPC Prediction and Cost Settings</h3>"),
    widgets.HTML(
        "<small>Q = tracking priority; R = voltage-change penalty; "
        "S = voltage-effort penalty.</small>"
    ),
    widgets.HBox([Np_w, Nc_w]),
    widgets.HBox([q_w, r_w, u_w])
])

classical_box = widgets.VBox([
    widgets.HTML("<h3>5. Classical Controller Gains</h3>"),
    widgets.HBox([p_kp_w]),
    widgets.HBox([pd_kp_w, pd_kd_w]),
    widgets.HBox([pi_kp_w, pi_ki_w]),
    widgets.HBox([pid_kp_w, pid_ki_w, pid_kd_w])
])

disturbance_box = widgets.VBox([
    widgets.HTML("<h3>6. External Load Disturbance</h3>"),
    enable_disturbance_w,
    widgets.HBox([disturbance_w, disturbance_time_w])
])

display(widgets.VBox([
    widgets.HTML(
        "<h2>DC Motor Speed Control Laboratory</h2>"
        "<h4>Comparison: P | PD | PI | PID | MPC</h4>"
        "<p>Configure the motor, reference speed, constraints and "
        "controller gains. Then run the simulations.</p>"
    ),

    motor_box,
    simulation_box,
    constraints_box,
    mpc_box,
    classical_box,
    disturbance_box,

    widgets.HBox([run_button, reset_button]),

    status_output,

    widgets.HTML("<h3>7. Simulation Plots</h3>"),
    plot_output,

    widgets.HTML("<h3>8. Performance Metrics and Model</h3>"),
    results_output,

    widgets.HTML("<h3>9. Export Results</h3>"),
    download_output
]))
```
